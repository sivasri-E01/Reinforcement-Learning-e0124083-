```python
import gymnasium as gym

env_names = [
    "CartPole-v1",
    "FrozenLake-v1",
    "MountainCar-v0",
    "Blackjack-v1",
    "Taxi-v4",          
    "CliffWalking-v1"
]

for name in env_names:
    print("=" * 60)
    print(f"ENVIRONMENT: {name}")
    print("=" * 60)

    env = gym.make(name)

    print(f"Observation (State) Space : {env.observation_space}")
    print(f"Action Space              : {env.action_space}")
    
    try:
        sample_state = env.observation_space.sample()
        sample_action = env.action_space.sample()

        P_dict = env.unwrapped.P

        if isinstance(sample_state, tuple):
            sample_state = list(P_dict.keys())[0]
            
        print(f"\nMDP Components Example for State {sample_state}:")
        
        for act in range(env.action_space.n):
            transitions = P_dict[sample_state][act]
            print(f"  Action {act}:")
            for prob, next_state, reward, done in transitions:
                print(f"    -> Prob: {prob} | Next State: {next_state} | Reward: {reward} | Terminated: {done}")
                
    except AttributeError:
        print("\nNote: This environment uses a continuous/fluid state space.")
        print("Transitions are computed dynamically via physics simulation instead of an explicit P table.")
        
    print("\n")
    env.close()
```
